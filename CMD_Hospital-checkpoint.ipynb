# CMS Hospital Performance Analysis
Analysis of 5,419 US hospitals using CMS public data — SQL, Excel, Power BI, and Python.

```python
import pandas as pd
from sqlalchemy import create_engine

engine = create_engine(
    "mssql+pyodbc://localhost\\SQLEXPRESS/CMS_Hospital_Data?"
    "driver=ODBC+Driver+17+for+SQL+Server&trusted_connection=yes"
)

df_tiers = pd.read_sql("SELECT * FROM vw_Hospital_Tiers", engine)
df_complications = pd.read_sql("SELECT * FROM vw_Hospital_Complications", engine)
df_readmissions = pd.read_sql("SELECT * FROM vw_Hospital_Readmissions", engine)

print(df_tiers.shape)
print(df_tiers.head())
```

## Correlation Analysis

```python
state_summary = pd.read_sql("SELECT * FROM vw_State_Summary", engine)
print(state_summary.head())

correlation = state_summary['Avg_Complication_Score'].corr(state_summary['Avg_Readmission_Score'])
print("Correlation between complications and readmissions:", correlation)
```

## Readmission Score by Performance Tier

```python
tier_summary = df_readmissions.groupby('Performance_Tier')['Readmission_Score'].mean()
print(tier_summary)
```

## Complications vs Readmissions — Scatter Plot with Trend Line

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("darkgrid")

plt.figure(figsize=(10, 6))
sns.regplot(
    data=state_summary,
    x='Avg_Complication_Score',
    y='Avg_Readmission_Score',
    scatter_kws={'color': '#60A5FA'},
    line_kws={'color': '#F87171'}
)
plt.title('Complications vs Readmissions by State (r = 0.66)', fontsize=14)
plt.xlabel('Average Complication Score')
plt.ylabel('Average Readmission Score')
plt.tight_layout()
plt.show()
```

## AMI Readmission Rate by Performance Tier — Boxplot

```python
ami_data = df_readmissions[df_readmissions['Readmission_Measure'] == 'Acute Myocardial Infarction (AMI) 30-Day Readmission Rate']

plt.figure(figsize=(10, 6))
sns.boxplot(
    data=ami_data,
    x='Performance_Tier',
    y='Readmission_Score',
    order=['High Performer', 'Average', 'Needs Improvement', 'Not Rated'],
    palette=['#34D399', '#FBBF24', '#F87171', '#9CA3AF']
)
plt.title('AMI Readmission Rate Distribution by Performance Tier', fontsize=14)
plt.xlabel('Performance Tier')
plt.ylabel('AMI Readmission Rate (%)')
plt.tight_layout()
plt.show()
```

## Complication & Readmission Scores by State — Heatmap

```python
from sklearn.preprocessing import MinMaxScaler

heatmap_data = state_summary.set_index('State')[['Avg_Complication_Score', 'Avg_Readmission_Score']]
heatmap_data_sorted = heatmap_data.sort_values('Avg_Readmission_Score', ascending=False)

scaler = MinMaxScaler()
scaled_for_color = heatmap_data_sorted.copy()
scaled_for_color[:] = scaler.fit_transform(heatmap_data_sorted)

plt.figure(figsize=(8, 14))
sns.heatmap(
    scaled_for_color,
    annot=heatmap_data_sorted,
    fmt='.1f',
    cmap='RdYlGn_r',
    linewidths=0.5,
    cbar_kws={'label': 'Relative Score (scaled per column)'}
)
plt.title('Complication & Readmission Scores by State (each column scaled independently)', fontsize=14)
plt.tight_layout()
plt.show()
```
